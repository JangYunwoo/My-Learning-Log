# RegEx

```python
import re
import numpy as np

# 1. 원본 비정형 데이터
raw_data = """
    sensor_1: 23.5 unit,
    sensor_2: 12.1 unit,
    sensor_3: 45.8 unit,
    sensor_4: error,
    sensor_5: 33.2 unit
"""

# 2. Regex 패턴 정의
# 숫자인 것(\d+), 소수점이 있을 수도 있는 것(\.?), 다시 숫자인 것(\d*)
pattern = re.compile(r"(\d+\.?\d*) unit")

# 3. 데이터 추출
# findall을 통해 'unit' 앞의 숫자만 리스트로 변환
extracted_values = pattern.findall(raw_data)

result = re.search(pattern, raw_data)
print(result.group())  # '23.5 unit'
print(result.group(1))  # '23.5'

print(f"추출된 텍스트 리스트: {extracted_values}")

# 4. Numpy로의 전환(Next Step)
# 리스트 형태의 문자열을 Numpy의 float64타입 배열로 변환
np_array = np.array(extracted_values, dtype=np.float64)
print(f"Numpy 배열 결과: {np_array}")
print(f"평균값 게산: {np_array.mean()}")
```

```python
text = "2026-02-20 ERROR user_id=12345 latency=203ms"

# 2-1, match / search
# match() : 문자열 시작부터 매칭, Match 객체 반환 (없으면 None)
print(re.match(r'\d{4}-\d{2}-\d{2}', text))
# <re.Match object; span=(0, 10), match='2026-02-20'>
print(re.match(r'ERROR', text))
# None (시작이 ERROR가 아님)

print(re.search(r'\d{4}-\d{2}-\d{2}', text))
# <re.Match object; span=(0, 10), match='2026-02-02'>
print(re.search(r'ERROR', text))
# <re.Match object; span=(11, 16), match='ERROR'>

# 2-2, findall / finditer
# findall() : 모든 매칭 결과를 리스트로 반환
print(re.findall(r'\d+', text))  # ['2026', '02', '20', '12345', '203']

# finditer() : 모든 매칭 결과를 Match 객체 이터레이터로 변환
for m in re.finditer(r'\d+', text):
    print(m.group(), m.span())
# 2026 (0, 4)
# 02 (5, 7)
# 20 (8, 10)
# 12345 (25, 30)
# 203 (39, 42)
```

```python
# 2-3. group
# .group() : 매칭된 문자열 반환
result = re.search(r'user_id=(\w+)', text)
print(result.group())  # 'user_id=12345' (전체)
print(result.group(1))  # '12345' (첫 번째 그룹 - 소괄호로 묶인 (\w+)를 의미)

# extra. 소괄호 사용법
# 소괄호의 개수보다 .group() 안의 숫자가 크면 error
extra_text = "2026-02-20 ERROR user_id=12345 latency=203ms"
extra_result = re.search(r'(\w+)=(\d+)', extra_text)
print(extra_result.group(0))  # 'user_id=12345' (전체)
print(extra_result.group(1))  # 'user_id' (첫 번째 그룹 - (\w+))
print(extra_result.group(2))  # '12345' (두 번째 그룹 - (\d+))

# .start() / .end() : 매칭된 위치의 시작/끝 인덱스
result = re.search(r'ERROR', text)
print(result.start())   # 11
print(result.end())     # 16

# .span() : (시작 인덱스, 끝 인덱스) 튜플 반환
print(result.span())    # (11, 16)

# 2-4. complie
# compile() : 패턴을 미리 컴파일하여 재사용
date_pattern = re.compile(r'\d{4}-\d{2}-\d{2}')

logs = ["2026-02-20 INFO start", "no date here", "2026-02-21 ERROR timeout"]
for log in logs:
    m = date_pattern.search(log)
    if m:
        print(m.group())
        # '2026-02-20'
        # '2026-02-21'
```

```python
file_list = [
    "report.pdf",
    "data.csv",
    "config.json",
    "image.png"
]

pattern = re.compile(r'\.(pdf|csv|json)$') # 여러 확장자를 하나로 묶음
for file in file_list:
    if pattern.search(file):
        print(f"Match: {file}")
        # Match: report.pdf
        # Match: data.csv
        # Match: config.json
```

```python
# 샘플 로그 (실제로는 수만 줄의 데이터일 수 있습니다)
log_data = "2026-02-23 14:30:05 [INFO] latency=156ms"

# 1. 인덱스 기반 그룹 추출
basic_pattern = re.compile(r"(\d{4}-\d{2}-\d{2}).*latency=(\d+)")
match = basic_pattern.search(log_data)

if match:
    print(f"전체 매치: {match.group()}") # 2026-02-23 14:30:05 [INFO] latency=156
    print(f"추출된 날짜: {match.group(1)}") # 2026-02-23
    print(f"추출된 지연시간: {match.group(2)}") # 156
```

```python
log_data = "2026-02-23 14:30:05 [INFO] latency=156ms"
named_pattern = re.compile(r"(?P<date>\d{4}-\d{2}-\d{2}).*latency=(?P<latency>\d+)")
match_named = named_pattern.search(log_data)

if match_named:
    print(f"전체 매치: {match_named.group(0)}") # 2026-02-03 14:30:05 [INFO] latency=156
    print(f"추출된 날짜: {match_named.group('date')}") # 2026-02-23
    print(f"추출된 지연시간: {match_named.group('latency')}") # 156
    # 딕셔너리 형태로 한 번에 변환도 가능(Pandas 데이터프레임 생성 시 유용)
    print(f"딕셔너리 변환: {match_named.groupdict()}")
    # {'date': '2026-02-23', 'latency': '156'}
```

```python
# 여러 데이터 매치
log_text = '''
2026-02-20 ERROR user_id=12345 msg="Login failed" latency=203ms
2026-02-21 INFO user_id=67890 latency=87ms
2026-02-22 WARN user_id=11111 msg="Long latency" latency=543ms
2026-02-23 ERROR user_id=22222 msg="Login failed" latency=1203ms
'''
named_pattern = re.compile(r"(?P<date>\d{4}-\d{2}-\d{2}).*latency=(?P<latency>\d+)")
match_list = named_pattern.finditer(log_text)
for match_data in match_list:
    print(match_data.groupdict())
    # {'date': '2026-02-20', 'latency': '203'}
    # {'date': '2026-02-21', 'latency': '87'}
    # {'date': '2026-02-22', 'latency': '543'}
    # {'date': '2026-02-23', 'latency': '1203'}
```

```python
html_text = """
<div>
    <p>Hello World</p>
</div>
"""

pattern = re.compile(r"<(?P<tag>\w+)>(.*?)</(?P=tag)>") # Named 예시
for match in pattern.finditer(html_text):
    print("Matched Text:", match.group(0)) # <p>Hello World</p>
    print("Tag Name:", match.group("tag")) # p
```

```python
# tag만 매칭하기
html = "<div>hello</div>"

print(re.findall(r"<.*>", html))  # ['<div>hello</div>'] - Greedy
print(re.findall(r"<.*?>", html)) # ['<div>', '</div>'] - Lazy
```

```python
text = "office: 02-123-4567"

# 1. 일반 그룹 () 사용 : 괄호 안의 내용(02)만 쏙 뽑아서 리스트에 담음
pattern1 = re.findall(r"(\d{2,3})-\d{3}-\d{4}", text)
print(f"일반 그룹: {pattern1}") # ['02']

# 2. 비캡처 그룹 (?:) 사용 : 괄호는 '묶기'용으로만 쓰고, 리스트에는 전체를 담음
pattern2 = re.findall(r"(?:\d{2,3})-\d{3}-\d{4}", text)
print(f"비캡처 그룹: {pattern2}") # ['02-123-4567']
```
